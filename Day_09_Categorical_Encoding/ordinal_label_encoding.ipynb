```python
import numpy as np 
import pandas as pd
df=pd.read_csv('/kaggle/input/datasets/nabinpaaudel/customer/customer.csv')
df.sample(5)
```

```python
df=df.iloc[:,2:]
df.head()
```

```python
X = df[['review', 'education']] 
y = df['purchased']
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[
    ['Poor', 'Average', 'Good'],  # Matches 'review' column
    ['School', 'UG', 'PG']         # Matches 'education' column
])
```

```python
oe.fit(X_train)
```

```python
X_train
```

```python
new_X_train=oe.transform(X_train)
new_X_test=oe.transform(X_test)
```
