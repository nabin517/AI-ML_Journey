```python
import numpy as  np 
flat_pixels=np.arange(12)
print(flat_pixels)
```

# 1. Reshaping and Transposing


```python
matrix_3x4=flat_pixels.reshape(3,4) # reshape 1D array into a 3x4 matrix
print('Reshaped to 3x4 Matrix:\n',matrix_3x4)
```

```python
trans_matrix=matrix_3x4.T # transpose a matrix
print('Transpose Matrix:\n',trans_matrix)
```

```python
flat_back=matrix_3x4.flatten()# flattened back a given matrix
print('Flattened Back to 1D:\n',flat_back)
```

# 2. Statistical Operation Along axes

```python
marks=np.array([
    [20,30,45],
    [50,60,75],
    [45,85,90]
])
print('Global median of marks',np.median(marks))
print('---------------------------------------------')
print('Global mean of marks',np.mean(marks))
print('---------------------------------------------')
print('Global standard deviation of marks',np.std(marks))
```

```python
print('Mean of marks [axis=0]',np.mean(marks,axis=0)) # gives the mean for each column 
print('------------------------------------------------------------------')
print('Mean of marks [axis=1]',np.mean(marks,axis=1)) # gives the mean for each row
```

```python
print('std of marks [axis=0]',np.std(marks,axis=0))
print('------------------------------------------------------------------')
print('std of marks [axis=0]',np.std(marks,axis=1))
```

```python
print('Location of max value is',np.argmax(marks)) # gives the location of  max_value
print('Location of min value is',np.argmin(marks)) # gives the location of  min_value
```

# 3.Matrix Multiplication

```python
X=np.array([
    [2,3,4],
    [1,3,5]
])
Y=np.array([
    [2,4],
    [1,5],
    [3,6]
])
```

```python
dot_result=np.dot(X,Y) # Traditional multiplication using dot_product
print(dot_result)
```

```python
matrix_mult=X@Y# Modern pythonic multiplication
print(matrix_mult)
```

# 4. Conditional Selection

```python
selection=np.where(marks>80)
print(selection)
```

```python
raw_layer_outputs = np.array([2.4, -1.5, 0.8, -3.2, 5.1])

# np.where(condition, value_if_true, value_if_false)
relu_outputs = np.where(raw_layer_outputs < 0, 0, raw_layer_outputs)
print("Raw Outputs:", raw_layer_outputs)
print("Activated Outputs (ReLU):", relu_outputs)
```
