# Complex Indexing(.loc[] and .iloc[])

```python
 print(df.loc[5,['Attack','Legendary']]) #df.loc[row_label, column_label]
```

```python
 print(df.loc[5,['Legendary']])
```

```python
print(df.loc[50:52,['Type 1','Type 2']])
```

```python
print(df.iloc[1,3]) # gives the first row fourth column
```

```python
print(df.iloc[0:2,2:4])
```

```python
print(df.iloc[:,7:9]) # gives all the rows of seventh and eight columns 
```

```python
print(df.iloc[103,:]) # gives all the columns of  row 103
```

```python
print(df.iloc[[103],:])# gives all the columns of row 103 as same orientation in the dataset .easier to look
```

# Groupby and Aggregation

```python
print(df.groupby('Type 1').mean(numeric_only=True))
```

```python
clean_summary=df.groupby('Type 2').agg(
    avg_hp=('HP','mean'),
    total_pokemon=('Name','count'),
    highest_attack=('Attack','max')
)
clean_summary
```
