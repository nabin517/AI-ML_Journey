```python
import pandas as pd 
import numpy as np 
df=pd.read_csv('/kaggle/input/datasets/nabinpaaudel/titanic-org/titanic.csv')
df.head()
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot(kind='bar')
fig.set_title('Passengers Travelling With')
```

```python
df['number_numerical']=pd.to_numeric(df['number'],downcast='integer',errors='coerce')
df['number_categorical']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
df.head()
```

```python
df['Cabin_num']=df['Cabin'].str.extract(r'(\d+)')
df['Cabin_cat']=df['Cabin'].str[0]
```

```python
df.sample(10)
```

```python
df['Ticket_num']=df['Ticket'].apply(lambda x:x.split()[-1])
df['Ticket_cat']=df['Ticket'].apply(lambda x:x.split()[0])
df['Ticket_cat']=np.where(df['Ticket_cat'].str.isdigit(),np.nan,df['Ticket_cat'])
```
