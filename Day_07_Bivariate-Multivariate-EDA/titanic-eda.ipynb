```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
titanic=pd.read_csv('/kaggle/input/datasets/yasserh/titanic-dataset/Titanic-Dataset.csv')
```

```python
titanic.head()
```

```python
cr_tab=pd.crosstab(titanic['Sex'],titanic['Survived'])
print(cr_tab)
```

# 1.(Categorical-Categorical) Plot

```python
plt.figure(figsize=(6,6))
sns.countplot(data=titanic,x='Sex',hue='Survived',palette='Set2')
plt.title('Survival Count By Sex vs Survived')
plt.legend(title='Survived',labels=['No','Yes'])
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.countplot(data=titanic,x='Pclass',hue='Survived')
plt.title('Survival Count By Pclass')
plt.legend(title="Survived",labels=['No','Yes'])
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.countplot(data=titanic,x='Embarked',hue='Survived')
plt.title('Port of Embarkation Vs Survived')
plt.legend(title='Survived',labels=['No','Yes'])
plt.show()
```

# 2.(Numerical-Categorical) Plot

```python
sns.boxplot(data=titanic,hue='Survived',y='Age',palette='Set1')
plt.title('Age Vs Survived')
plt.show()
```

```python
sns.boxplot(data=titanic,hue='Survived',y='Fare',palette='Set2')
plt.title('Fare Vs Survived')
plt.legend(title='Survived',labels=['No','Yes'])
plt.show()
```

# 3.MultiVariate Analysis

```python
sns.barplot(data=titanic,x='Pclass',y='Survived',hue='Sex',errorbar=None)
plt.title('Survival Rate By Class and Gender Together')
plt.show()
```

```python
sns.scatterplot(data=titanic,x='Age',y='Fare',hue='Survived',palette='Set2')
plt.title('Age Vs Fare Vs Survived')
plt.show()
```

```python
sns.scatterplot(data=titanic,x='SibSp',y='Fare',hue='Survived',palette='Set1')
plt.title('Sibsp Vs Fare Vs Survived')
plt.show()
```
