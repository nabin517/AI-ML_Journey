```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y,color="red", linestyle='--', marker="*", label="y=2x")
plt.title("My first Plot")
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()# legend displays the label  look at in the top left corner 
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [22, 24, 19, 23, 25]
humidity = [30, 35, 40, 32, 31]
```

```python
plt.plot(temperature,days,color='blue',linestyle='--',marker='*',label='temperature')
plt.plot(humidity,days,color='Green',linestyle='-',marker='o',label='Humidity')
plt.xlabel('Temperature')
plt.ylabel('Days')
plt.legend()
plt.show()
```

```python
import pandas as pd
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [200, 250, 300, 280, 350]
}
df = pd.DataFrame(data)
plt.plot(df["Month"], df["Sales"],marker='o' )
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

# BarChart

```python
regions=['Nepal','India','China','Bhutan','Srilanka']
sales=[300,250,400,450,500]
bar=plt.bar(regions,sales,color=['blue','red','yellow'])
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
regions = ["North", "South", "East", "West"]
sales = [300, 250, 400, 350]
fig, ax = plt.subplots()
bars = ax.bar(regions, sales)
ax.bar_label(bars)
plt.title("Sales by Region")
plt.show()
```

# Histplot

```python
import seaborn as sns
scores=[12,23,56,78,65,56,13,78,56,43,43,12,43,43]
sns.histplot(x=scores,bins=5,kde=True,binwidth=3)
plt.title('Score Distribution')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.set_theme(style="whitegrid") 
sns.histplot(x=scores, binwidth=5, kde=True, color='royalblue', edgecolor='white')
plt.title('Distribution of Test Scores')
plt.xlabel('Scores Obtained')
plt.ylabel('Number of Students')
plt.axvline(80, color='crimson', linestyle='--', label='Passing Mark (80)')
plt.legend()
plt.show()
```

```python
days = [1, 2, 3, 4]
accuracy = [65, 78, 85, 92]
plt.plot(days, accuracy, label='Model Accuracy')
plt.title('Model Training Progress over Days')
plt.xlabel('Days of Training')
plt.ylabel('Accuracy Percentage (%)')
plt.legend()
plt.show()
```
