```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv('/kaggle/input/datasets/dilmurodgoyupov/weight-height/weight-height.csv')
df.head()
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(df['Height'],kde=True)

plt.subplot(1,2,2)
sns.histplot(df['Weight'],kde=True)
plt.show()
```

```python
df['Height'].describe()
```

```python
sns.boxplot(x=df['Height'])
plt.show()
```

```python
upper_limit=df['Height'].quantile(0.99)
lower_limit=df['Height'].quantile(0.01)
```

```python
upper_limit,lower_limit
```

```python
new_df=df[(df['Height']>=58.13) & (df['Height']<=74.78)]
```

```python
new_df['Height'].describe()
```

```python
sns.histplot(new_df['Height'],kde=True)
plt.show()
```

```python
sns.boxplot(x=new_df['Height'])
plt.show()
```

# Capping

```python
df['Height']=np.where(df['Height']>=upper_limit,upper_limit,np.where(df['Height']<=lower_limit,lower_limit,df['Height']))
```
