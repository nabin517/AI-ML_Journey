```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings
warnings.filterwarnings('ignore')
np.random.seed(42)
```

```python
data=pd.DataFrame({
    'Square_ft':np.random.normal(loc=2000,scale=500,size=100),
    'Room':np.random.choice([2,3,4,5],size=100),
    'House-Type':np.random.choice(['Apartment','Villa'],size=100)
})
```

```python
data['Price']=data['Square_ft']*0.5+np.random.normal(0,30,size=100)
```

# 1.Scatter Plot


```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=data,x='Square_ft',y='Price',hue='House-Type')
plt.title('House Price vs Square Feet')
plt.xlabel('Square Feet')
plt.ylabel('Price')
plt.show()
```

# 2.Boxplot and Violin Plot

```python
fig,axes=plt.subplots(1,2,figsize=(12,5))
sns.boxplot(data=data,x='Room',y='Price',ax=axes[0],palette='Set2')
axes[0].set_title('Price distribution By Room')
sns.violinplot(data=data,x='Room',y='Price',ax=axes[1],palette='Set2')
s=axes[1].set_title('Price distribution By Room')
plt.tight_layout()
plt.show()
```

# 3.HeatMap and Correlation Matrix

```python
plt.figure(figsize=(6,4))
numeric_data=data.drop(columns=['House-Type'])
correlation_matrix=numeric_data.corr()
sns.heatmap(correlation_matrix,annot=True,vmin=-1,vmax=1,cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

# 4.Pairplot

```python
sns.pairplot(data=data, hue='House-Type', palette='Dark2')
plt.suptitle('Complete Dataset Pair Plot Matrix', y=1.02)
plt.show()
```
