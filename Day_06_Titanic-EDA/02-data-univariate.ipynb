```python
import pandas as pd
import warnings
import seaborn as sns
df=pd.read_csv('/kaggle/input/datasets/nabinpaaudel/traind/train.csv')
df.head(2)
```

# 1.Categorical Data 

# a.countplot

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
df['Pclass'].value_counts().plot(kind='bar')
```

# b.Piechart

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

# 2.Numerical data

# a.Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'],bins=50) # bins can be used for small range or large range
```

# b.Distplot

```python
sns.displot(df['Age'],kde=True) # kde is used to find the distribution of the data
warnings.filterwarnings('ignore')
```

# c.Boxplot

```python
sns.boxplot(df['Age'])
```
